```python
import pandas as pd
import numpy as np
import os
import random
from datetime import datetime, timedelta

# This tells Python to go "up one level" and then into 'data_inputs'
input_dir = '../data_inputs'
os.makedirs(input_dir, exist_ok=True)

brokers = ['Apex_Securities', 'Beacon_Trades', 'Summit_Clearing', 'Vanguard_Feeds', 'Horizon_Brokers']
symbols = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA']

def generate_broker_data(broker_name, num_rows=1000):
    data = []
    start_date = datetime.now() - timedelta(days=1)
    
    for i in range(num_rows):
        trade_id = f"TRD-{broker_name[:3].upper()}-{10000 + i}"
        symbol = random.choice(symbols)
        quantity = random.randint(10, 500)
        
        # Inject occasional errors intentionally!
        rand_val = random.random()
        
        if rand_val < 0.02:  # 2% chance of missing price (Null value)
            price = np.nan
        elif rand_val < 0.03:  # 1% chance of an insane price anomaly
            price = random.choice([-50.0, 999999.99])
        else:
            price = round(random.uniform(50.0, 500.0), 2)
            
        trade_time = start_date + timedelta(seconds=random.randint(0, 86400))
        
        data.append({
            'TradeID': trade_id,
            'Broker': broker_name,
            'Symbol': symbol,
            'Quantity': quantity,
            'Price': price,
            'TradeTimestamp': trade_time.strftime('%Y-%m-%d %H:%M:%S')
        })
    
    df = pd.DataFrame(data)
    
    # Inject a few duplicate TradeIDs on purpose
    if random.random() > 0.5:
        duplicates = df.sample(n=5).copy()
        duplicates['Quantity'] = duplicates['Quantity'] * 2 
        df = pd.concat([df, duplicates], ignore_index=True)
        
    # Save to CSV
    file_path = f"{input_dir}/{broker_name}_trades.csv"
    df.to_csv(file_path, index=False)
    print(f"✅ Success: Generated {len(df)} rows for {broker_name}")

# Run the generator
for b in brokers:
    generate_broker_data(b)
```
